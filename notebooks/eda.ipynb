# Boston housing EDA

Quick exploration of the dataset before deciding on feature engineering.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_boston

raw = load_boston()
df = pd.DataFrame(raw.data, columns=raw.feature_names)
df['MEDV'] = raw.target
df.shape
```

```python
df.describe()
```

## Distributions

CRIM and LSTAT are heavily right-skewed. Good candidates for log1p.

```python
fig, axes = plt.subplots(4, 4, figsize=(14, 10))
for ax, col in zip(axes.ravel(), df.columns):
    ax.hist(df[col], bins=30)
    ax.set_title(col)
plt.tight_layout()
```

## Correlations with MEDV

RM and LSTAT are the strongest predictors (positive and negative respectively).

```python
corr = df.corr()
corr['MEDV'].sort_values()
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm')
plt.title('correlation matrix')
```

```python
# scatter of the two strongest predictors against the target
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].scatter(df['RM'], df['MEDV'], alpha=0.4)
ax[0].set_xlabel('RM')
ax[0].set_ylabel('MEDV')
ax[1].scatter(df['LSTAT'], df['MEDV'], alpha=0.4)
ax[1].set_xlabel('LSTAT')
ax[1].set_ylabel('MEDV')
plt.tight_layout()
```

## Notes

- LSTAT vs MEDV looks roughly hyperbolic: log transform helps.
- RM vs MEDV looks roughly linear with a quadratic curve at the extremes.
- B is extremely concentrated near the upper bound; not informative. Drop it.
